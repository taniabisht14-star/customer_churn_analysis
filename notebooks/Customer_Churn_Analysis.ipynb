```python
#create first purchase 
customer_master["first_purchase_month"] = (
    customer_master["first_purchase_date"]
    .dt.to_period("M")
)
```

```python
customer_master[
    ["customer_id", "first_purchase_date", "first_purchase_month"]
].head()
```

```python
#add purchase month to valid orders
valid_orders["purchase_month"] = (
    valid_orders["order_date"]
    .dt.to_period("M")
)
```

```python
#attach each custom to their orders
valid_orders_cohort = valid_orders.merge(
    customer_master[
        ["customer_id", "first_purchase_month"]
    ],
    on="customer_id",
    how="left"
)
```

```python
#calculate months since first purchase 
valid_orders_cohort["months_since_first_purchase"] = (
    valid_orders_cohort["purchase_month"]
    - valid_orders_cohort["first_purchase_month"]
).apply(lambda x: x.n)
```

```python
#create chort customer counts 
cohort_retention = (
    valid_orders_cohort
    .groupby(
        [
            "first_purchase_month",
            "months_since_first_purchase"
        ]
    )["customer_id"]
    .nunique()
    .reset_index()
)

cohort_retention.head(10)
```

```python
# size
cohort_sizes = (
    customer_master[
        customer_master["total_orders"] > 0
    ]
    .groupby("first_purchase_month")["customer_id"]
    .nunique()
    .reset_index()
)

cohort_sizes.columns = [
    "first_purchase_month",
    "cohort_size"
]
```

```python
#calculate retention percentage 
cohort_retention = cohort_retention.merge(
    cohort_sizes,
    on="first_purchase_month",
    how="left"
)

cohort_retention["retention_rate"] = (
    cohort_retention["customer_id"]
    / cohort_retention["cohort_size"]
    * 100
)
```

```python
cohort_retention.head(15)
```

```python
#create retention matrix
retention_matrix = cohort_retention.pivot(
    index="first_purchase_month",
    columns="months_since_first_purchase",
    values="retention_rate"
)

retention_matrix
```

```python
#categorical churn analysis
categorical_drivers = [
    "gender",
    "region",
    "city_tier",
    "acquisition_channel",
    "loyalty_member",
    "preferred_device"
]

for column in categorical_drivers:
    print("\n" + "=" * 60)
    print(column.upper())
    
    analysis = (
        customer_master
        .groupby(column)
        .agg(
            customers=("customer_id", "nunique"),
            churned_customers=(
                "customer_status",
                lambda x: (x == "Churned").sum()
            )
        )
        .reset_index()
    )
    
    analysis["churn_rate"] = (
        analysis["churned_customers"]
        / analysis["customers"]
        * 100
    )
    
    print(analysis.sort_values("churn_rate", ascending=False))
```

```python
# creating a purchasing-customer analysis
purchasing_customers = customer_master[
    customer_master["total_orders"] > 0
].copy()

print("Purchasing customers:", len(purchasing_customers))
```

```python
#analyze each categorical data 
for column in categorical_drivers:
    print("\n" + "=" * 60)
    print(column.upper())
    
    analysis = (
        purchasing_customers
        .groupby(column)
        .agg(
            customers=("customer_id", "nunique"),
            churned_customers=(
                "customer_status",
                lambda x: (x == "Churned").sum()
            )
        )
        .reset_index()
    )
    
    analysis["churn_rate"] = (
        analysis["churned_customers"]
        / analysis["customers"]
        * 100
    )
    
    print(
        analysis
        .sort_values("churn_rate", ascending=False)
        .round(2)
    )
```

```python
#delivery delay as a churn driver 
purchasing_customers["delivery_delay_band"] = pd.cut(
    purchasing_customers["delivery_delay_rate"],
    bins=[-0.01, 10, 25, 50, 100],
    labels=[
        "0-10%",
        "10-25%",
        "25-50%",
        "50%+"
    ]
)
```

```python
delivery_driver = (
    purchasing_customers
    .groupby("delivery_delay_band", observed=False)
    .agg(
        customers=("customer_id", "nunique"),
        churned_customers=(
            "customer_status",
            lambda x: (x == "Churned").sum()
        )
    )
    .reset_index()
)

delivery_driver["churn_rate"] = (
    delivery_driver["churned_customers"]
    / delivery_driver["customers"]
    * 100
)

delivery_driver.round(2)
```

```python
#marketing engagement vs churn 
purchasing_customers["marketing_engaged"] = np.where(
    purchasing_customers["open_rate"] > 0,
    "Engaged",
    "Not Engaged"
)
```

```python
marketing_driver = (
    purchasing_customers
    .groupby("marketing_engaged")
    .agg(
        customers=("customer_id", "nunique"),
        churned_customers=(
            "customer_status",
            lambda x: (x == "Churned").sum()
        )
    )
    .reset_index()
)

marketing_driver["churn_rate"] = (
    marketing_driver["churned_customers"]
    / marketing_driver["customers"]
    * 100
)

marketing_driver.round(2)
```

```python
#At-Risk Customer Identification
```

```python
#create risk segemnts
purchasing_customers["risk_segment"] = np.select(
    [
        purchasing_customers["days_since_last_purchase"] >= 60,
        purchasing_customers["days_since_last_purchase"] >= 30
    ],
    [
        "High Risk",
        "Medium Risk"
    ],
    default="Low Risk"
)
```

```python
#create active at-risk customers
at_risk_customers = purchasing_customers[
    purchasing_customers["customer_status"] == "Active"
].copy()

at_risk_customers = at_risk_customers[
    at_risk_customers["days_since_last_purchase"] >= 30
]
```

```python
at_risk_customers[
    [
        "customer_id",
        "customer_name",
        "days_since_last_purchase",
        "total_orders",
        "historical_clv",
        "delivery_delay_rate",
        "open_rate",
        "risk_segment"
    ]
].head(10)
```

```python
#find high value at-risk customers 
high_value_at_risk = at_risk_customers[
    at_risk_customers["clv_segment"] == "High Value"
].sort_values(
    "historical_clv",
    ascending=False
)
```

```python
high_value_at_risk[
    [
        "customer_id",
        "customer_name",
        "days_since_last_purchase",
        "total_orders",
        "historical_clv",
        "delivery_delay_rate",
        "open_rate",
        "risk_segment"
    ]
].head(20)
```

```python
#summarize the risk population 
risk_summary = (
    at_risk_customers
    .groupby("risk_segment")
    .agg(
        customers=("customer_id", "nunique"),
        total_historical_value=("historical_clv", "sum"),
        average_clv=("historical_clv", "mean"),
        average_days_since_purchase=(
            "days_since_last_purchase",
            "mean"
        )
    )
    .reset_index()
)

risk_summary
```

```python
#revenue at Risk & Win-Back Opportunity
```

```python
#calcualte value active at-risk customers
at_risk_value = (
    at_risk_customers
    .groupby("risk_segment")
    .agg(
        customers=("customer_id", "nunique"),
        historical_value=("historical_clv", "sum"),
        average_customer_value=("historical_clv", "mean")
    )
    .reset_index()
)

at_risk_value
```

```python
#calclate churned customer valye 
churned_customers = purchasing_customers[
    purchasing_customers["customer_status"] == "Churned"
].copy()

churned_value = (
    churned_customers
    .agg(
        churned_customers=("customer_id", "nunique"),
        total_churned_value=("historical_clv", "sum"),
        average_churned_value=("historical_clv", "mean")
    )
)

churned_value
```

```python
#create win back priority
at_risk_customers["winback_priority"] = np.select(
    [
        (
            (at_risk_customers["risk_segment"] == "High Risk") &
            (at_risk_customers["clv_segment"] == "High Value")
        ),
        (
            (at_risk_customers["risk_segment"] == "High Risk") &
            (at_risk_customers["clv_segment"] == "Medium Value")
        ),
        (
            (at_risk_customers["risk_segment"] == "Medium Risk") &
            (at_risk_customers["clv_segment"] == "High Value")
        )
    ],
    [
        "Critical",
        "High",
        "High"
    ],
    default="Standard"
)
```

```python
#see the priority distribution 
at_risk_customers["winback_priority"].value_counts()
```

```python
#create final list
winback_list = at_risk_customers[
    at_risk_customers["winback_priority"].isin(
        ["Critical", "High"]
    )
].sort_values(
    ["winback_priority", "historical_clv"],
    ascending=[True, False]
)
```

```python
winback_list[
    [
        "customer_id",
        "customer_name",
        "days_since_last_purchase",
        "total_orders",
        "historical_clv",
        "delivery_delay_rate",
        "open_rate",
        "conversion_rate",
        "risk_segment",
        "clv_segment",
        "winback_priority"
    ]
].head(20)
```

```python
#calculate total value of priority customer s
priority_value = (
    winback_list["historical_clv"].sum()
)

priority_customers = (
    winback_list["customer_id"].nunique()
)

print("Priority customers:", priority_customers)
print("Historical value:", round(priority_value, 2))
```
